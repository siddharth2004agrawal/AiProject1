# Make a class Simple agent

```python
class simpleAgent:
  def __init__(self,name="Basic Agent"):
    self.name=name
    self.current_task=None
    self.completed_tasks=[]
    self.knowledge_base=[]

```

```python
def perceive(self,environment_state):
  print(f"{self.name}perceives:{environment_state}")

```

```python
def plan(self,goal):
  print(f"{self.name} is planning to achieve: {goal}")
  if "make coffe" in goal.lower():
    self.current_task=[
        "Get coffee beans",
        "Grind beans",
        "Boil water",
        "Brew coffee",
        "Pour into cup"
    ]
  elif"write report" in goal.lower():
    self.current_task=[
        "Gather data",
        "outline Report",
        "Draft Section",
        "Review and edit",
        "Finalize Report"
    ]
  else:
    self.current_task=[goal]
    print(f"{self.name}planned steps:{self.current_task}")
    return
```

```python
def act(self):
  if not self.current_task:\
  print(f"{self,name}:No current task to act upon")
  return
  if self.current_task:
    task=self.current_task.pop(0)
    print(f"{self.name}is acting:{action}")
    import time
    time.sleep(0.5)
    self.completed_tasks.append(action)
    return f"completed'{action})'"
    return "No action taken."
```

```python
def reflect(self, outcome):
  print(f"{self.name} is reflecting on outcome:{outcome}")
  self.knowledge_base.append(outcome)
  if not self.current_task:
    print(f"{self.name} All pllanned steps for the current goal are completed.Reflecting on overall success.")
  else
  print(f"{self.name} still:{len(self.knowledge_base)}")
  return
```
