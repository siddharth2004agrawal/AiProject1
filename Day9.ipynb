# Load ML and Python Library

```python
from sklearn.datasets import make_classification
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
```

```python
X_waste, y_waste = make_classification(
    n_samples=1000, n_features=10, n_informative=5, n_redundant=2,
    n_classes=3, n_clusters_per_class=1, random_state=42 )

print(f"Synthetic waste data shape: {X_waste.shape}")
print(f"Synthetic waste target shape: {y_waste.shape}")
print(f"Number of samples: {len(X_waste)}")
print(f"Number of classes: {len(np.unique(y_waste))}")

print("\nFirst 5 rows of features (X):")
print(X_waste[:5])
print("\nFirst 5 rows of targets (y):")
print(y_waste[:5])

X_train_waste,X_test_waste,y_train_waste,y_test_waste = train_test_split(X_waste, y_waste, test_size=0.2, random_state=42)

scaler_waste = StandardScaler()
X_train_waste_scaled = scaler_waste.fit_transform(X_train_waste)
X_test_waste_scaled = scaler_waste.transform(X_test_waste)
print(f"Training data shape after scaling: {X_train_waste_scaled.shape}")
print(f"Test data shape after scaling: {X_test_waste_scaled.shape}")

model_waste = LogisticRegression(max_iter=1000, solver='liblinear', random_state=42)
model_waste.fit(X_train_waste_scaled, y_train_waste)

print("Waste classification model training complete!")
y_pred_waste = model_waste.predict(X_test_waste_scaled)

accuracy_waste = accuracy_score(y_test_waste, y_pred_waste)
print(f"\nAccuracy: {accuracy_waste:.4f}")

print("\nClassification Report:")
print(classification_report(y_test_waste, y_pred_waste))

cm_waste = confusion_matrix(y_test_waste, y_pred_waste)
plt.figure(figsize=(8, 6))
sns.heatmap(cm_waste, annot=True, fmt='d', cmap='Blues', cbar=False,
            xticklabels=[f'Class {i}' for i in np.unique(y_waste)],
            yticklabels=[f'Class {i}' for i in np.unique(y_waste)])
plt.xlabel('Predicted Label')
plt.ylabel('True Label')
plt.title('Confusion Matrix for Waste Classification')
plt.show()
```
