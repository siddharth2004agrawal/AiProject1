# Simple Agentic AI Example: Task Completion Agent

```python
import time
import pandas as pd

class SimpleAgent:
    def __init__(self, name="Basic Agent"):
        self.name = name
        self.current_task = None
        self.completed_tasks = []
        self.knowledge_base = []

    def perceive(self, environment_state):
        print(f"{self.name} perceives: {environment_state}")

    def plan(self, goal):
        print(f"{self.name} is planning to achieve: {goal}")
        if "make coffee" in goal.lower():
            self.current_task = [
                "Get coffee beans",
                "Grind beans",
                "Boil water",
                "Brew coffee",
                "Pour into cup"
            ]
        elif "write report" in goal.lower():
            self.current_task = [
                "Gather data",
                "Outline report",
                "Draft sections",
                "Review and edit",
                "Finalize report"
            ]
        else:
            self.current_task = [goal]
        print(f"{self.name} planned steps: {self.current_task}")

    def act(self):
        if not self.current_task:
            print(f"{self.name}: No current task to act upon.")
            return None

        action = self.current_task.pop(0)
        print(f"{self.name} is acting: {action}")
        time.sleep(0.1) # Reduced simulation sleep time for faster execution
        self.completed_tasks.append(action)
        return f"Completed '{action}'"

    def reflect(self, outcome):
        print(f"{self.name} reflects on outcome: {outcome}")
        self.knowledge_base.append(outcome)
        if not self.current_task:
            print(f"{self.name}: All planned steps completed. Reflecting on success.")
        else:
            print(f"{self.name}: Still {len(self.current_task)} steps remaining. Continuing...")

# --- 1. Coffee Maker Agent ---
agent = SimpleAgent("Coffee Maker Agent")
goal = "Make coffee"
agent.perceive("Kitchen is ready for coffee making.")
agent.plan(goal)

while agent.current_task:
    result = agent.act()
    agent.reflect(result)

print(f"\n{agent.name} has completed all tasks for goal: '{goal}'")
print(f"Final completed tasks: {agent.completed_tasks}")
print(f"Agent's knowledge base: {agent.knowledge_base}\n")

# --- 2. Report Writing Agent ---
agent_report = SimpleAgent("Report Writing Agent")
goal_report = "Write a detailed report"
agent_report.perceive("Research materials are available.")
agent_report.plan(goal_report)

while agent_report.current_task:
    result_report = agent_report.act()
    agent_report.reflect(result_report)

print(f"\n{agent_report.name} has completed all tasks for goal: '{goal_report}'")
print(f"Final completed tasks: {agent_report.completed_tasks}")
print(f"Agent's knowledge base: {agent_report.knowledge_base}\n")

# --- 3. Save & Verify Data ---
# FIX: Instantiate the missing Pandas Series from the first agent's knowledge base
knowledge_series = pd.Series(agent.knowledge_base, name="Knowledge Base")

csv_filename = 'coffee_maker_agent_knowledge_base.csv'
knowledge_series.to_csv(csv_filename, index=False)
print(f"Agent's knowledge base saved to '{csv_filename}'\n")

# Display the content of the saved CSV to verify
print("CSV File Header Preview:")
print(pd.read_csv(csv_filename).head())

report_knowledge_series = pd.Series(agent_report.knowledge_base, name='completed_action')
# Save the Series to a CSV file
report_csv_filename = 'report_writing_agent_knowledge_base.csv'
report_knowledge_series.to_csv(report_csv_filename, index=False)
print(f"'Report Writing Agent's knowledge base saved to '{report_csv_filename}'")
```
