# Simple Rule-based Chatbot Example

```python
def simple_chatbot():
  print("Hello! i am a simple chatbot.Type 'quit'to exit.")
  while True:
    user_input=input("You: ").lower()
    if user_input=="quit":
      print("Chatbot:Goodbye!")
      break
    elif 'hello'in user_input or 'hi' in user_input:
        print("Chatbot:Hi there! How can I help you today?")
    elif 'how are you'in user_input:
        print("Chatbot:I'm just a program,but I'am doing great!Thanks for asking.")
    elif 'name'in user_input:
        print("Chatbot:I don't have a name.You can call me Chatbot.!")
    elif 'weather' in user_input:
      print("Chatbot:I cannnot tell you about the weather,as i don't have access to real-time data.")
    elif 'help' in user_input:
      print("Chatbot:I can answer basic questions or just chat try asking me about my name!")
    elif 'who is your creater'in user_input:
      print("Chatbot: Siddharth Agrawal is your creater..!")
    else:
      print("chatbot:I'm sorry,I don't understand that. Can you rephrashe?")
```

```python
simple_chatbot()
```

# Simulate a chat conversation history

```python
conversation_history = [ "Chatbot: Hello! I'm a simple chatbot. Type 'quit' to exit.",
                        "You: hello", "Chatbot: Hi there! How can I help you today?",
                         "You: how are you",
                         "Chatbot: I'm just a program, but I'm doing great! Thanks for asking.",
                         "You: what is your name", "Chatbot: I don't have a name. You can call me Chatbot!",
                         "You: quit",
                         "Chatbot: Goodbye!"
                         ]
```

Save the conversation to the text file

```python
filename = "chatbot_conversation.txt"
try:
  with open(filename, "w") as f:
    for line in conversation_history:
     f.write(line + "\n")
    print(f"Conversation successfully saved to {filename}")
except IOError as e:
    print(f"Error saving conversation to file: {e}")
```

Read and display the content of the saved file

```python
try:
  with open(filename, "r") as f:
   content = f.read()
  print("\n--- Contents of " + filename + " ---\n")
  print(content)
  print("\n----------------------------------")
except FileNotFoundError: print(f"Error: The file {filename} was not found.")
except IOError as e: print(f"Error reading file: {e}")
```

```python
new_chat_entries = [
    "You: Tell me a joke.",
    "Chatbot: Why don't scientists trust atoms? Because they make up everything!",
    "You: That's a good one!"
    ]
try:
    with open(filename, "a") as f:
     for entry in new_chat_entries:
      f.write(entry + "\n")
      print(f"Successfully appended new entries to {filename}")
except IOError as e: print(f"Error appending to file: {e}")

try:
  with open(filename, "r") as f:
    updated_content = f.read()
    print("\n--- Updated Contents of " + filename + " ---\n")
    print(updated_content)
    print("\n----------------------------------------")
except FileNotFoundError:
  print(f"Error: The file {filename} was not found.")
except IOError as e: print(f"Error reading file: {e}")
```
